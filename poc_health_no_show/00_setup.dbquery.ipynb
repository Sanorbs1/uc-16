```sql
CREATE CATALOG IF NOT EXISTS poc_health_no_show;

CREATE SCHEMA IF NOT EXISTS poc_health_no_show.bronze;
CREATE SCHEMA IF NOT EXISTS poc_health_no_show.silver;
CREATE SCHEMA IF NOT EXISTS poc_health_no_show.gold;
CREATE SCHEMA IF NOT EXISTS poc_health_no_show.ops;

COMMENT ON CATALOG poc_health_no_show IS
'POC for appointment no-show risk identification and outreach drafting. Non-production demo environment.';


--Create landing volume
CREATE VOLUME IF NOT EXISTS poc_health_no_show.bronze.landing;


CREATE TABLE IF NOT EXISTS poc_health_no_show.ops.pipeline_run_audit (
  run_id STRING,
  pipeline_name STRING,
  task_name STRING,
  status STRING,
  start_time TIMESTAMP,
  end_time TIMESTAMP,
  rows_processed BIGINT,
  error_message STRING
)
COMMENT 'Operational audit table for UC-16 pipeline runs.';


CREATE TABLE IF NOT EXISTS poc_health_no_show.ops.data_quality_results (
  run_id STRING COMMENT 'Unique monitoring run ID',
  check_name STRING COMMENT 'Name of data-quality check',
  table_name STRING COMMENT 'Table being checked',
  status STRING COMMENT 'PASS or FAIL',
  failed_record_count BIGINT COMMENT 'Number of failing records',
  total_record_count BIGINT COMMENT 'Total records checked',
  checked_at TIMESTAMP COMMENT 'Check execution time'
)
COMMENT 'Data-quality monitoring results for UC-16.';
CREATE TABLE IF NOT EXISTS poc_health_no_show.ops.ai_message_quality_results (
  run_id STRING,
  appointment_id STRING,
  variant STRING,
  check_name STRING,
  status STRING,
  message_length INT,
  checked_at TIMESTAMP
)
COMMENT 'Quality checks for AI-generated outreach messages.';

```
