```python
# Databricks notebook source
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

spark.sql("""
CREATE OR REPLACE TABLE poc_health_no_show.gold.appointment_monthly_trend
COMMENT 'Monthly appointment and no-show trend for operational monitoring.'
AS
SELECT
  date_trunc('month', appointment_date) AS appointment_month,
  COUNT(*) AS appointment_count,
  SUM(CASE WHEN no_show_flag THEN 1 ELSE 0 END) AS no_show_count,
  ROUND(
    SUM(CASE WHEN no_show_flag THEN 1 ELSE 0 END) / COUNT(*) * 100,
    2
  ) AS no_show_rate_pct
FROM poc_health_no_show.gold.appointment_risk_segments
GROUP BY date_trunc('month', appointment_date)
ORDER BY appointment_month
""")

print("Monthly no-show trend table created successfully.")
```
