```python
# Databricks notebook source
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

spark.sql("""
CREATE OR REPLACE TABLE poc_health_no_show.gold.outreach_priority_queue
COMMENT 'Ranked outreach queue for high and medium-high risk appointments.'
AS
SELECT
  appointment_id,
  patient_pseudo_id,
  appointment_date,
  appointment_weekday,
  risk_segment,
  risk_score,
  risk_reason,
  recommended_action,
  row_number() OVER (
    PARTITION BY appointment_date
    ORDER BY risk_score DESC, prior_no_show_rate DESC
  ) AS outreach_priority_rank
FROM poc_health_no_show.gold.appointment_risk_segments
WHERE risk_segment IN ('HIGH', 'MEDIUM_HIGH')
""")

print("Outreach priority queue created successfully.")
```
