```python
# Databricks notebook source
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

spark.sql("""
CREATE OR REPLACE TABLE poc_health_no_show.gold.appointment_risk_enriched
COMMENT 'Appointment risk data enriched with business-friendly age and lead-time buckets.'
AS
SELECT
  *,
  CASE
    WHEN age < 18 THEN 'Under 18'
    WHEN age BETWEEN 18 AND 35 THEN '18-35'
    WHEN age BETWEEN 36 AND 55 THEN '36-55'
    WHEN age BETWEEN 56 AND 75 THEN '56-75'
    ELSE '76+'
  END AS age_group,
  CASE
    WHEN lead_time_days <= 0 THEN 'Same day'
    WHEN lead_time_days BETWEEN 1 AND 2 THEN '1-2 days'
    WHEN lead_time_days BETWEEN 3 AND 7 THEN '3-7 days'
    WHEN lead_time_days BETWEEN 8 AND 14 THEN '8-14 days'
    ELSE '15+ days'
  END AS lead_time_bucket
FROM poc_health_no_show.gold.appointment_risk_segments
""")

print("Appointment risk enrichment table created successfully.")
```
